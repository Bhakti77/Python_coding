```
# Write a function that checks whether a given number is prime.
# 🧠 A prime number is a number greater than 1 that has no divisors other than 1 and itself.

def is_prime(n):
    if n <= 1:
        print("should be greater than 1")
        return False
    else:
        for x in range(2,n):
            if n%x == 0: # it is divisible by others
                return False
            return True
        

ans = is_prime(7)
ans2 = is_prime(200)
print(ans)
print(ans2)
```
